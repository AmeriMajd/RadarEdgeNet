```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import os
demo_folder = r"C:\Users\ameri\MmWave\Radars_Fusion\demo"
os.makedirs(demo_folder, exist_ok=True)
```

```python
# 1. Raw Data Scatterplot
np.random.seed(42)
raw_x = np.random.normal(0, 10, 300)
raw_y = np.random.normal(0, 10, 300)

plt.figure(figsize=(6,5))
plt.scatter(raw_x, raw_y, c="blue", alpha=0.6, s=20)
plt.title("Raw Radar Data")
plt.xlabel("x_cc")
plt.ylabel("y_cc")
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig(os.path.join(demo_folder, "01_raw_data.png"))
plt.close()
```

```python
# 2. Preprocessing (highlight selected sensors)
sensor_mask = raw_x > 0
plt.figure(figsize=(6,5))
plt.scatter(raw_x[~sensor_mask], raw_y[~sensor_mask], c="lightgray", alpha=0.4, s=20, label="Other sensors")
plt.scatter(raw_x[sensor_mask], raw_y[sensor_mask], c="green", alpha=0.7, s=30, label="Selected sensors (1&2)")
plt.title("Preprocessing: Selected Sensors")
plt.xlabel("x_cc")
plt.ylabel("y_cc")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig(os.path.join(demo_folder, "02_selected_sensors.png"))
plt.close()
```

```python
# 3. Track Summarization (one track)
track_x = np.random.normal(5, 1, 30)
track_y = np.random.normal(3, 1, 30)
centroid = (np.mean(track_x), np.mean(track_y))

plt.figure(figsize=(6,5))
plt.scatter(track_x, track_y, c="orange", s=40, alpha=0.7, label="Track detections")
plt.scatter(*centroid, c="red", marker="x", s=100, label="Summarized centroid")
plt.title("Track Summarization")
plt.xlabel("x_cc")
plt.ylabel("y_cc")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig(os.path.join(demo_folder, "03_track_summarization.png"))
plt.close()
```

```python
# 4. Sensor Fusion (two groups matched)
sensor1 = np.random.normal([-3, 0], 1, (20, 2))
sensor2 = np.random.normal([3, 0], 1, (20, 2))
pairs = list(zip(sensor1[:10], sensor2[:10]))

plt.figure(figsize=(6,5))
plt.scatter(sensor1[:,0], sensor1[:,1], c="blue", label="Sensor 1")
plt.scatter(sensor2[:,0], sensor2[:,1], c="green", label="Sensor 2")
for (p1, p2) in pairs:
    plt.plot([p1[0], p2[0]], [p1[1], p2[1]], "k--", alpha=0.5)
plt.title("Sensor Fusion: Matched Objects")
plt.xlabel("x_cc")
plt.ylabel("y_cc")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig(os.path.join(demo_folder, "04_sensor_fusion.png"))
plt.close()

```

```python
# 5. Class distribution (imbalanced dataset)
labels = ["Pedestrian", "Cyclist", "Other"]
counts = [180, 230, 5000]

plt.figure(figsize=(6,5))
plt.bar(labels, counts, color=["red","orange","gray"])
plt.title("Class Distribution (Before Augmentation)")
plt.ylabel("Count")
plt.savefig(os.path.join(demo_folder, "05_class_distribution.png"))
plt.close()
```

```python
# 6. After Augmentation (scatter with extra points)
minority_x = np.random.normal(0, 1, 50)
minority_y = np.random.normal(0, 1, 50)
aug_x = minority_x + np.random.normal(0, 0.2, 50)
aug_y = minority_y + np.random.normal(0, 0.2, 50)

plt.figure(figsize=(6,5))
plt.scatter(minority_x, minority_y, c="red", label="Original Minority Class")
plt.scatter(aug_x, aug_y, c="pink", alpha=0.6, label="Augmented Samples")
plt.title("Data Augmentation (Pedestrian Example)")
plt.xlabel("x_cc")
plt.ylabel("y_cc")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.savefig(os.path.join(demo_folder, "06_augmentation.png"))
plt.close()
```

```python
# 7. SMOTE (balanced dataset bar chart)
balanced_counts = [5000, 5000, 5000]
plt.figure(figsize=(6,5))
plt.bar(labels, balanced_counts, color=["red","orange","gray"])
plt.title("Class Distribution (After SMOTE)")
plt.ylabel("Count")
plt.savefig(os.path.join(demo_folder, "07_smote.png"))
plt.close()
```

```python
# 8. Feature importances (dummy values)
features = ["x_cc", "y_cc", "vr_mean", "x_std", "y_std", "rcs_mean", "width", "num_points", "abs_vr_mean"]
importances = np.random.dirichlet(np.ones(len(features)), size=1).flatten()

plt.figure(figsize=(8,5))
plt.barh(features, importances, color="skyblue")
plt.title("Random Forest Feature Importances")
plt.xlabel("Importance")
plt.gca().invert_yaxis()
plt.savefig(os.path.join(demo_folder, "08_feature_importances.png"))
plt.close()
```

```python
# 9. Confusion Matrix (dummy)
conf_matrix = np.array([[80, 5, 15],
                        [10, 70, 20],
                        [25, 15, 460]])

fig, ax = plt.subplots(figsize=(6,5))
cax = ax.matshow(conf_matrix, cmap="Blues")
plt.title("Confusion Matrix", pad=20)
fig.colorbar(cax)
for (i,j), val in np.ndenumerate(conf_matrix):
    ax.text(j, i, f"{val}", va="center", ha="center")
ax.set_xticks(range(3))
ax.set_yticks(range(3))
ax.set_xticklabels(labels)
ax.set_yticklabels(labels)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.savefig(os.path.join(demo_folder, "09_confusion_matrix.png"))
plt.close()
```

```python
# 10. Two Radar Detection Fusion Visualization

# Create synthetic data for two radar sensors
np.random.seed(42)

# Sensor 1 detections
sensor1_x = np.random.normal(-2, 1.5, 25)
sensor1_y = np.random.normal(2, 1.5, 25)
sensor1_ids = [f"S1_{i}" for i in range(len(sensor1_x))]

# Sensor 2 detections (similar but not identical to sensor1)
sensor2_x = np.random.normal(2, 1.5, 25)
sensor2_y = np.random.normal(-2, 1.5, 25)
sensor2_ids = [f"S2_{i}" for i in range(len(sensor2_x))]

# Create some matching pairs between sensors
matched_pairs = [(0, 0), (1, 2), (3, 1), (5, 4), (7, 6), (8, 9), (10, 11)]

plt.figure(figsize=(10, 8))

# Plot sensor 1 detections
plt.scatter(sensor1_x, sensor1_y, c='blue', s=100, marker='o', alpha=0.8,
            edgecolors='black', linewidth=1.5, label='Sensor 1 Detections')

# Plot sensor 2 detections
plt.scatter(sensor2_x, sensor2_y, c='red', s=100, marker='s', alpha=0.8,
            edgecolors='black', linewidth=1.5, label='Sensor 2 Detections')

# Draw lines between matched pairs and plot fused points
for idx1, idx2 in matched_pairs:
    x1, y1 = sensor1_x[idx1], sensor1_y[idx1]
    x2, y2 = sensor2_x[idx2], sensor2_y[idx2]

    # Draw dashed line between sensors
    plt.plot([x1, x2], [y1, y2], 'k--', alpha=0.7, linewidth=1)

    # Plot fused point (midpoint with special marker)
    fused_x, fused_y = (x1 + x2) / 2, (y1 + y2) / 2
    plt.scatter(fused_x, fused_y, color='green', s=200, marker='*', alpha=1.0,
                edgecolors='gold', linewidth=2)

    # Add text showing the fusion
    plt.text(fused_x + 0.3, fused_y + 0.3, f"Fused\n(S1_{idx1}+S2_{idx2})",
             fontsize=9, ha='left', va='center',
             bbox=dict(boxstyle="round,pad=0.3", facecolor='lightyellow', alpha=0.8))

plt.title("Radar Fusion: Matching Objects Between Two Sensors")
plt.xlabel("x_cc (meters)")
plt.ylabel("y_cc (meters)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.axis('equal')
plt.savefig(os.path.join(demo_folder, "10_radar_fusion_matching.png"), dpi=150, bbox_inches='tight')
plt.close()

# 11. Step-by-Step Fusion Process

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Radar Fusion Process Step-by-Step', fontsize=16, fontweight='bold')

# Panel 1: Sensor 1 detections only
axes[0].scatter(sensor1_x, sensor1_y, c='blue', s=100, marker='o', alpha=0.8)
for i, (x, y) in enumerate(zip(sensor1_x, sensor1_y)):
    axes[0].text(x + 0.2, y + 0.2, f"S1_{i}", fontsize=9, ha='left')
axes[0].set_title("Sensor 1 Detections")
axes[0].set_xlabel("x_cc (meters)")
axes[0].set_ylabel("y_cc (meters)")
axes[0].grid(True, alpha=0.3)
axes[0].axis('equal')

# Panel 2: Sensor 2 detections only
axes[1].scatter(sensor2_x, sensor2_y, c='red', s=100, marker='s', alpha=0.8)
for i, (x, y) in enumerate(zip(sensor2_x, sensor2_y)):
    axes[1].text(x + 0.2, y + 0.2, f"S2_{i}", fontsize=9, ha='left')
axes[1].set_title("Sensor 2 Detections")
axes[1].set_xlabel("x_cc (meters)")
axes[1].set_ylabel("y_cc (meters)")
axes[1].grid(True, alpha=0.3)
axes[1].axis('equal')

# Panel 3: Fused objects
axes[2].scatter(sensor1_x, sensor1_y, c='blue', s=60, marker='o', alpha=0.5)
axes[2].scatter(sensor2_x, sensor2_y, c='red', s=60, marker='s', alpha=0.5)

# Draw lines between matched pairs and plot fused points
for idx1, idx2 in matched_pairs:
    x1, y1 = sensor1_x[idx1], sensor1_y[idx1]
    x2, y2 = sensor2_x[idx2], sensor2_y[idx2]

    # Draw line between sensors
    axes[2].plot([x1, x2], [y1, y2], 'k--', alpha=0.5, linewidth=1)

    # Plot fused point (midpoint)
    fused_x, fused_y = (x1 + x2) / 2, (y1 + y2) / 2
    axes[2].scatter(fused_x, fused_y, color='green', s=150, marker='*', alpha=1.0)
    axes[2].text(fused_x + 0.3, fused_y + 0.3, f"Fused\n(S1_{idx1}+S2_{idx2})",
                 fontsize=8, ha='left', va='center')

axes[2].set_title("Fused Objects")
axes[2].set_xlabel("x_cc (meters)")
axes[2].set_ylabel("y_cc (meters)")
axes[2].grid(True, alpha=0.3)
axes[2].axis('equal')

plt.tight_layout()
plt.savefig(os.path.join(demo_folder, "11_fusion_process_step_by_step.png"), dpi=150, bbox_inches='tight')
plt.close()

# 12. Fusion Benefits - Improved Object Localization

# Create data showing how fusion improves localization accuracy
true_x, true_y = 0, 0  # True object position

# Sensor 1 measurements with error
s1_measurements_x = np.random.normal(-2, 1.2, 20)
s1_measurements_y = np.random.normal(2, 1.2, 20)

# Sensor 2 measurements with error
s2_measurements_x = np.random.normal(2, 1.2, 20)
s2_measurements_y = np.random.normal(-2, 1.2, 20)

# Fused measurements (average)
fused_x = (s1_measurements_x + s2_measurements_x) / 2
fused_y = (s1_measurements_y + s2_measurements_y) / 2

plt.figure(figsize=(10, 8))

# Plot sensor 1 measurements
plt.scatter(s1_measurements_x, s1_measurements_y, c='blue', s=50, alpha=0.6,
            label='Sensor 1 Measurements')

# Plot sensor 2 measurements
plt.scatter(s2_measurements_x, s2_measurements_y, c='red', s=50, alpha=0.6,
            label='Sensor 2 Measurements')

# Plot fused measurements
plt.scatter(fused_x, fused_y, c='green', s=80, alpha=0.8,
            label='Fused Measurements')

# Plot true position
plt.scatter(true_x, true_y, c='black', s=200, marker='X',
            label='True Position')

# Add confidence ellipses
from matplotlib.patches import Ellipse


def plot_confidence_ellipse(ax, x, y, color, n_std=2.0):
    cov = np.cov(x, y)
    pearson = cov[0, 1] / np.sqrt(cov[0, 0] * cov[1, 1])
    ell_radius_x = np.sqrt(1 + pearson)
    ell_radius_y = np.sqrt(1 - pearson)
    ellipse = Ellipse((0, 0), width=ell_radius_x * 2, height=ell_radius_y * 2,
                      facecolor='none', edgecolor=color, linewidth=2, alpha=0.7)

    scale_x = np.sqrt(cov[0, 0]) * n_std
    mean_x = np.mean(x)
    scale_y = np.sqrt(cov[1, 1]) * n_std
    mean_y = np.mean(y)

    ellipse.set_center((mean_x, mean_y))
    ellipse.width = scale_x * 2
    ellipse.height = scale_y * 2
    ax.add_patch(ellipse)


plot_confidence_ellipse(plt.gca(), s1_measurements_x, s1_measurements_y, 'blue')
plot_confidence_ellipse(plt.gca(), s2_measurements_x, s2_measurements_y, 'red')
plot_confidence_ellipse(plt.gca(), fused_x, fused_y, 'green')

plt.title("Fusion Benefits: Improved Localization Accuracy")
plt.xlabel("x_cc (meters)")
plt.ylabel("y_cc (meters)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.axis('equal')
plt.savefig(os.path.join(demo_folder, "12_fusion_benefits.png"), dpi=150, bbox_inches='tight')
plt.close()

print("Radar fusion visualizations created:")
print("10_radar_fusion_matching.png - Shows object matching between two radar sensors")
print("11_fusion_process_step_by_step.png - Step-by-step fusion process")
print("12_fusion_benefits.png - Demonstrates improved localization accuracy through fusion")
```
